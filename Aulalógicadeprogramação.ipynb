```python
%%bash
# Save the C code to a file
cat << EOF > program.c
#include <stdio.h>
#include <stdlib.h>
int main() /* Explicitly declare return type as int */
{
    char tamanho;
    printf ("\n Digite o tamanho [G para grande, M para medio ou P para pequeno]");
    scanf (" %c", &tamanho); /* Added space before %c for robust input reading */
    if (tamanho == 'G' || tamanho == 'g')
        printf ("\n www.tamanhog.com.br \n\n");
    else if (tamanho == 'M' || tamanho == 'm')
        printf ("\n www.tamanhom.com.br \n\n");
    else if (tamanho == 'P' || tamanho == 'p')
        printf ("\n www.tamanhop.com.br \n\n");
    else
        printf ("\n Entrada invalida. Por favor, digite G, M ou P.\n\n");
    return 0; /* Return 0 to indicate successful execution */
}
EOF

# Compile the C code
gcc program.c -o program

# Run the compiled program with 'G' input
echo "G" | ./program

echo "---------------------"

# Run the compiled program with 'M' input
echo "M" | ./program

echo "---------------------"

# Run the compiled program with 'P' input
echo "P" | ./program

echo "---------------------"

# Run the compiled program with an invalid input
echo "X" | ./program
```
